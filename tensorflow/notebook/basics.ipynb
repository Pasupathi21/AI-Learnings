```python
import tensorflow as tf
```

```python
# scalar - single value shape will be 0

scalar = tf.constant(2, dtype=float)

print(scalar)
```

```python
# vector - list of values one dimension only have shape no range
vector = tf.constant([1, 2, 3, 4])
print(vector)
```

```python
# matrix - 2 dimension 
matrix = tf.constant([
    [1, 2, 3, 4], 
    [4, 3, 2, 1]])
print(matrix)
```

```python
# tensor - 3 dimension row, column & range(direction)
# below exmaple 
# block - 2 
# row - 3
# colum - 2
# shape of the tensor is 2, 3, 2
tensor = tf.constant([
    [
        [1, 2],
        [3, 4],
        [4, 5]
    ],
    [
        [5, 6],
        [7, 8],
        [8, 9]
    ]
])
print(tensor)
```

```python
import numpy as np
```

```python
# convert tensor to numpy vice versa

new_tensor = tf.constant([
    [
        [
            [1, 2]
        ]
    ],
    [
        [
            [1, 2]
        ]
    ],
    [
        [
            [1, 2]
        ]
    ]
])

print(new_tensor.numpy())

```

```python
# convert numpy to tensor 
num_list = np.array([1, 2, 4, 5])
print(num_list)
print(tf.constant(num_list))
```
